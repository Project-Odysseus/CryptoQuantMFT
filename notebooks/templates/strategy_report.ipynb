# Strategy report

The standard report for a finished candidate, and the decision that follows. The report's first section is written
for a portfolio manager; the rest is the evidence.

````python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from quantlab.lab import Lab
from quantlab.signals import library as lib
from quantlab.reporting import charts

charts.apply_style()
pd.set_option("display.width", 200)

from IPython.display import HTML, display

lab = Lab(universe="equity_index_core", start="1999", show=False)
idea = lab.signal(lib.trend_forecast("compound_sigmoid"), name="smooth trend (H13 design)")
````

````python
path = lab.report(idea, definition="A volatility-normalised EMA difference through a compound sigmoid (Martin 2023, H13), "
                                   "equal-weighted over the seven core equity indices at 10% volatility each, trading at the next close.")
display(HTML(f'<a href="{path.resolve().as_uri()}" target="_blank">open the report</a>'))
````
